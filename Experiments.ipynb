## Plot Probability of A winning against difference in realsied load, deltaL

## Plot of Probability of A winning against number of contests. Varying amounts of error shown with 

```python
from scipy.stats import binom
import math
p = 1 / (1 + math.exp(-0.2))
m = 99
P = 1 - binom.cdf(m//2, m, p)
print(P)

plt.axhline(P, color='gray', linestyle='--', label='y = 10')
# Plot_prob_against_m: (pop_size, num_trials, req_deltaL, error)
y, x = Population.Plot_prob_against_m(500, 10000, 0.2, 0)
y2, x2 = Population.Plot_prob_against_m(500, 10000, 0.2, 0.2)
y3, x3 = Population.Plot_prob_against_m(500, 10000, 0.2, 0.5)
y4, x4 = Population.Plot_prob_against_m(500, 10000, 0.2, 1)
plt.scatter(x, y, label='Error = 0')
plt.scatter(x2, y2, label='Error = 0.2')
plt.scatter(x3, y3, label='Error = 0.5')
plt.scatter(x4, y4, label='Error = 1')
plt.xlabel('Number of contests, m')
plt.ylabel('Probability of A winning')
plt.legend()
plt.show()
```
